```python



def performance_metrics(
    users:int,
    requests_per_second: int,
    requests_per_second_peak: int,
    request_processing_time: float,
    extended_processing_time: float,
    available_connections: int,
    user_bit_rate: int,
    file_size:float,
    line_rate:float,
    peak_percentage: float,
    hours_in_peak: float
):

    U = users
    R = line_rate
    lambda_ = requests_per_second
    lambda_peak = requests_per_second_peak
    T = request_processing_time
    Tex = extended_processing_time
    C = available_connections
    B = user_bit_rate
    F = file_size

    streaming =  lambda_ * T
    streming_bandwidth = U * B 
    total_volume = U * F
    download_time = ( U * F ) / R
    l_max = C / Tex
    saturation = streaming > L_max
    request_increse_at_saturation = lambda_ - l_max
    peak_factor = (peak_percentage * 24) / hours_in_peak

    return {
        "streaming": streaming,
        "streming_bandwidth": streming_bandwidth,
        "total_volume": total_volume,
        "download_time": download_time,
        "l_max": l_max,
        "saturation": saturation,
        "request_increse_at_saturation": request_increse_at_saturation,
        "peak_factor": peak_factor
    }
    

diagnostics = performance_metrics(
    users: 8000,
    requests_per_second: int,
    requests_per_second_peak: int,
    request_processing_time: float,
    extended_processing_time: float,
    available_connections: int,
    user_bit_rate: int,
    file_size:float,
    line_rate:float,
    peak_percentage: float,
    hours_in_peak: float
)

diagnostics
```
