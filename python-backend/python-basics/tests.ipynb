```python
calls = [
    {"id": 1, "facility": "F1", "status": "answered"},
    {"id": 2, "facility": "F1", "status": "missed"},
    {"id": 3, "facility": "F2", "status": "missed"},
    {"id": 4, "facility": "F2", "status": "missed"},
]
counts = {}
for c in calls:
    counts[c["status"]] = counts.get(c["status"], 0) + 1
print(counts)
```

```python
def connect(host, port, timeout):
    return f"Connecting to {host}:{port} with timeout {timeout}s"

config = {"host": "localhost", "port": 5432, "timeout": 30}
result = connect(**config)
# Equivalente a: connect(host="localhost", port=5432, timeout=30)
print(result)  # "Connecting to localhost:5432 with timeout 30s"
```
